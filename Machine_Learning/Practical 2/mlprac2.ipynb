```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('sales_data_sample.csv',encoding='unicode_escape')
df.head()
```

```python
df.info()
```

```python
df_drop  = ['ADDRESSLINE1', 'ADDRESSLINE2', 'POSTALCODE', 'CITY', 'TERRITORY', 'PHONE', 'STATE', 'CONTACTFIRSTNAME', 'CONTACTLASTNAME', 'CUSTOMERNAME', 'ORDERNUMBER']
df = df.drop(df_drop, axis=1)
```

```python
df.info()
```

```python
for col in df.columns.values:
    print(df[col].value_counts())
```

```python
df.drop(columns=['ORDERDATE','STATUS','MONTH_ID','QTR_ID','YEAR_ID'],inplace=True)
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
def convert_categories(col):
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].values)
```

```python
categories = ['PRODUCTLINE','PRODUCTCODE','COUNTRY','DEALSIZE']
for col in categories:
    convert_categories(col)
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
data = sc.fit_transform(df)
```

```python
import os
os.environ["OMP_NUM_THREADS"] = "12"
```

```python
from sklearn.cluster import KMeans
import warnings

warnings.filterwarnings(
    "ignore",
    message="KMeans is known to have a memory leak on Windows with MKL.*",
    category=UserWarning
)

wcss = []

for k in range(1, 15):
    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        random_state=15,
        n_init=10
    )
    kmeans.fit(data)
    wcss.append(kmeans.inertia_)

print(wcss)
```

```python
k = list(range(1,15))
plt.plot(k,wcss)
plt.xlabel('Clusters')
plt.ylabel('scores')
plt.title('Finding right number of clusters')
plt.grid()
plt.show()
```

```python

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Create 4 clusters
kmeans = KMeans(n_clusters=4, init='k-means++', random_state=15, n_init=10)
labels = kmeans.fit_predict(data)

# Plot clustered dots
plt.figure(figsize=(8, 6))

plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis', s=50)

# Plot centroids
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    c='red', marker='X', s=250, label='Centroids'
)

plt.title('K-Means Clustering of Dots')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
